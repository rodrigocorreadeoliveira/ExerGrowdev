```python
# Importar bibliotecas
!pip install pandas graphviz
import pandas as pd
from graphviz import Digraph
```

```python
# Definir as entidades com seus respectivos atributos
# Cada entidade terá uma lista de atributos
entidades = {
    'Produto': ['ID_Produto', 'Nome', 'Descrição', 'Preço', 'Estoque', 'ID_Categoria'],
    'Categoria': ['ID_Categoria', 'Nome', 'Descrição'],
    'Cliente': ['ID_Cliente', 'Nome', 'Endereço', 'Telefone', 'Email', 'Data_Nascimento'],
    'Pedido': ['ID_Pedido', 'Data_Pedido', 'Status', 'Valor_Total', 'ID_Cliente'],
    'Item_do_Pedido': ['ID_Item', 'Quantidade', 'Preço_Unitário', 'Subtotal', 'ID_Pedido', 'ID_Produto'],
    'Pagamento': ['ID_Pagamento', 'Data_Pagamento', 'Valor_Pago', 'Método_Pagamento', 'ID_Pedido']
}
```

```python
# Criar o diagrama ER usando a biblioteca Graphviz
# Inicializa o gráfico
diagrama_er = Digraph('Diagrama_ER', format='png')
diagrama_er.attr(rankdir='LR')  # Define a orientação do gráfico
```

```python
# Adicionar entidades ao diagrama
# Loop pelas entidades, adicionando cada uma ao diagrama
for entidade, atributos in entidades.items():
    label = f'{entidade}|' + '|'.join(atributos)
    diagrama_er.node(entidade, shape='record', label=label)
```

```python
# Definir os relacionamentos
relacionamentos = [
    ('Produto', 'Categoria', 'Pertence a'),
    ('Cliente', 'Pedido', 'Faz'),
    ('Pedido', 'Item_do_Pedido', 'Contém'),
    ('Pedido', 'Pagamento', 'Possui')
]
```

```python
# Adicionar os relacionamentos
for origem, destino, relacionamento in relacionamentos:
    diagrama_er.edge(origem, destino, label=relacionamento)
```

```python
# Renderizar o diagrama ER e salvar o arquivo
diagrama_er.render('ecommerce_er_diagram', view=True)
```
